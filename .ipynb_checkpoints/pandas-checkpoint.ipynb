### **Python Pandas**

```python
import pandas as pd
```

### **Pandas Series**

```python
carname = pd.Series(["BMW","KIA","TATA","NISSAN"])
carname
```

```python
print(carname[0])
```

```python
print(carname.dtype)
print(carname.shape)
print(carname.size)
```

```python
carname.isna().sum()
```

```python
carname.info()
```

```python
carcolour = pd.Series(["Red","Blue","Black","Grey"])
carcolour
```

```python
cardata = pd.DataFrame({
    "Name":carname,
    "Colour":carcolour
})
cardata
```

```python
print(cardata.dtypes)
print("------------")
print(cardata.shape)
print("------------")
print(cardata.size)
```

```python
cardata.info()
```
