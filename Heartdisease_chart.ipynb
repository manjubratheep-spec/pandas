```python
import pandas as pd
```

```python
df = pd.read_csv("D:\\Gen AI\\Car_Sales_CSV\\heart-disease.csv")
df
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(df['age'], df['trestbps'],c=df["target"],cmap="viridis")
plt.xlabel("Age")
plt.ylabel("trestbps")
plt.title("Trestbps Levels")
plt.axhline(df['trestbps'].mean(), color='r', linestyle='dashed',label="BP")
plt.legend()
plt.colorbar(label="Target")
plt.show();
```

```python
plt.scatter(df[df["target"] == 0]["age"],df[df["target"] == 0]["cp"],label="Target 0")
plt.scatter(df[df["target"] == 1]["age"],df[df["target"] == 1]["cp"],label="Target 1")
plt.xlabel("Age")
plt.ylabel("cp")
plt.title("cp pain Levels")

plt.axhline(df['cp'].mean(), color='r', linestyle='dashed', label="cp avarage")

plt.legend()
plt.show()
```
