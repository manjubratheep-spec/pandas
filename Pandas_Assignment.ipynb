### **4.Find the mean, maximum, minimum, and median salary. Also find the average age of all employees.**

```python
print("Mean Salary:", df["Salary"].mean())
print("Maximum Salary:", df["Salary"].max())
print("Minimum Salary:", df["Salary"].min())
print("Median Salary:", df["Salary"].median())
print()
print("Average age of all employees")
print("Average Age:", df["Age"].mean())
```

### **5.For each department, calculate the number of employees, average salary, and highest salary. Sort the result by average salary in descending order.**

```python
df1 = df.groupby("Department").agg(
    EmployeeCount=("EmpID", "count"),
    AverageSalary=("Salary", "mean"),
    HighestSalary=("Salary", "max"))
df1 = df1.sort_values(by="AverageSalary", ascending=False)
df1
```

### **8.Create a column named SalaryLevel: 'High' for salary greater than or equal to 50,000, 'Medium' for salary from 40,000 to 49,999, and 'Low' for salary below 40,000. Display Name, Salary, and SalaryLevel.**

```python
def salary_level(salary):
    if salary >= 50000:
        return "High"
    elif salary >= 40000:
        return "Medium"
    else:
        return "Low"

df["SalaryLevel"] = df["Salary"].apply(salary_level)
df[["Name", "Salary", "SalaryLevel"]]
```

### **9.Create a marks DataFrame with columns EmpID and TrainingScore, containing scores for EmpID 101, 103, 105, 108, and 110. Merge it with the employee DataFrame using EmpID. Display all employees and their scores, including employees without a score.**

```python
marks = pd.DataFrame({
    "EmpID": [101, 103, 105, 108, 110],
    "TrainingScore": [70, 90, 65, 80, 75]
})
rlt = pd.merge(df, marks, on="EmpID", how="left")

rlt[["EmpID", "Name", "TrainingScore"]]
```

### **10.Find the department with the highest average salary. Then display the employees whose salary is above the overall average salary. Write 2–3 sentences explaining your findings.**

```python
avg_salary = df.groupby("Department")["Salary"].mean()
print("Highest Average Salary Department:", avg_salary.idxmax())
```

```python
overall_avg = df["Salary"].mean()
print("Overall Average Salary:", overall_avg)
```

```python
rlt1 = df[df["Salary"] > overall_avg]
rlt1[["Name", "Department", "Salary"]]
```

##### ***#10 explanation***

* df.groupby("Department")["Salary"].mean() groups the DataFrame by the "Department" column and calculates the average (mean) salary for each department.
* .idxmax() finds and returns the index label (which is the department name) that has the maximum average salary value.
* df["Salary"].mean() computes the average salary across all employees in the entire dataset, regardless of their department.
* df[df["Salary"] > overall_avg] creates a filtered subset of the DataFrame (rlt1) containing only the rows where the employee's salary is strictly greater than the overall average (44400.0).
* rlt1[["Name", "Department", "Salary"]] selects and displays only specific columns (Name, Department, and Salary) for those filtered employees.
