```python
import matplotlib.pyplot as plt
```

```python
plt.figure()
plt.plot()
plt.show();
```

### **Scatter plot**

```python
import numpy as np

x = np.array([1,2,3])
y = np.array([2,3,1])
plt.scatter(x,y,c=('r','b','y'),marker="*",s=150)
plt.show();
```

### **Line plot**

```python
x = np.array([1,2,3,4])
y = np.array([2,3,1,4])
c = ['red','yellow','pink','blue']
plt.plot(x,y,c='g',marker='*',linestyle = 'dashed')
plt.scatter(x,y,color= c,marker="*",s=150)
plt.show();
```

### **Bar chart**

```python
students = ["Arun", "Bala", "Charan", "Divya", "Eswar"]
marks = [85, 70, 90, 65, 78]
c = ['red','yellow','green','blue','pink']
plt.bar(students,marks,color=c,align='edge',edgecolor='black')
plt.title('student marks')
plt.show();
```

```python
# frquency distribution

np.random.seed(2)
a = np.random.normal(500,100,1000)
plt.hist(a,color='olive',bins=20,edgecolor="black")
plt.show();
```

### **Pie chart**

```python
activities = ["Sleep", "Study", "Work", "Food", "Exercise"]
hours = [8, 5, 6, 2, 3]
exp = [0.08,0,0,0,0]
c = ['pink','red','olive','yellow','blue']
plt.pie(hours,labels=activities,colors= c,explode=exp)
plt.title('Daily activities')
plt.show();
```

```python
# Stack market
stockA = [1,2,3,4,5]
stockB = [3,4,5,3,2]
stockC = [6,4,2,1]
plt.title("Stack market")
plt.plot(stockA,label="Stock1")
plt.plot(stockB,label="Stock2")
plt.plot(stockC,label="Stock3")
plt.legend()
plt.show();
```
