```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("D:\\Gen AI\\failed_ssh_attempts.csv")
df
```

```python
top_10 = df.sort_values( by="Failed_Login_Count", ascending=False).head(10)
top_10
```

### **Bar Chart**

```python
plt.figure(figsize=(10, 5))

plt.bar(top_10["IP_Address"], top_10["Failed_Login_Count"],color='skyblue',edgecolor="black")

plt.xlabel("IP Address")
plt.ylabel("Failed SSH Login Attempts")
plt.title("Top 10 IP Addresses - Failed SSH Login Attempts")
plt.xticks(rotation=89)
#plt.show();

plt.tight_layout()
```
